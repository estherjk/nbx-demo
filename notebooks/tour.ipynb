# nbx tour

This is a real `.ipynb` file, open as one Python buffer. Each border line is a cell.

- Put the cursor in a cell and press **Shift+Enter** to run it and move on.
- `\ra` runs every cell, `\R` restarts the kernel, `\i` interrupts it.
- `]c` and `[c` jump between cells.

```python
import sys

print("hello from Python", sys.version.split()[0])
2 ** 10
```

## Output arrives as it is produced

Run this one and watch the lines appear. Press `\i` to interrupt it halfway.

```python
import time

for step in range(1, 6):
    print(f"step {step}/5")
    time.sleep(0.5)
```

## Plots

Images are drawn inline in Rex, Ghostty and kitty. Elsewhere you get a text placeholder.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)
plt.plot(x, np.sin(x), label="sin")
plt.plot(x, np.cos(x), label="cos")
plt.legend()
plt.show()
```

## Editing cells

- `\a` adds a code cell below, `\M` a markdown cell.
- `\m` turns this cell into code and back.
- `\J` and `\K` move a cell, `\d` deletes it.
- `:w` saves source and outputs to the notebook.

State carries over between cells, as in any notebook:

```python
def reciprocal(n):
    return 1 / n


x.shape, reciprocal(4)
```

## Errors

Tracebacks keep their colours. When output is long, `\o` opens all of it in a split.

A kernel stops at the first error, so with `\ra` any cells after a failing one are skipped.

```python
reciprocal(0)
```
